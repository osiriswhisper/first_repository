```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv('/content/train.csv')
test_df = pd.read_csv('/content/test.csv')
```

```python
df.info()
```

df.drop(['Name', 'Cabin', 'Embarked', 'Ticket'], axis=1, inplace=True)
test_df.drop(['Name', 'Cabin', 'Embarked', 'Ticket'], axis=1, inplace=True)

```python
# Create a binary feature: 1 if cabin info exists, 0 otherwise
df['Has_Cabin'] = df['Cabin'].notna().astype(int)
test_df['Has_Cabin'] = test_df['Cabin'].notna().astype(int)
```

```python
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
test_df['Sex'] = test_df['Sex'].map({'male': 0, 'female': 1})
```

```python
df['Embarked']=df['Embarked'].map({'S':0,'C':1,'Q':2})
test_df['Embarked']=test_df['Embarked'].map({'S':0,'C':1,'Q':2})
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy='median')
df[['Age']] = imputer.fit_transform(df[['Age']])
test_df[['Age']] = imputer.transform(test_df[['Age']])
test_df[['Fare']] = imputer.fit_transform(test_df[['Fare']])
```

```python
features = ['Pclass', 'Sex', 'Age','SibSp']
X = df[features]
y = df['Survived']
X_test_final = test_df[features]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2, random_state=23)
```

```python
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
val_pred = model.predict(X_val)
print("Validation Accuracy:", accuracy_score(y_val, val_pred))
```

```python
final_preds = model.predict(X_test_final)
```

```python
submission = pd.DataFrame({
    'PassengerId': test_df['PassengerId'],
    'Survived': final_preds
})
```

```python
submission.to_csv('submission6.csv', index=False)
```
